```python
import os
import matplotlib.pyplot as plt
from collections import defaultdict
```

```python
def get_plant_distribution(dataset_path):
    plant_counts = defaultdict(int)
    total_images = 0
    
    # Traverse the dataset directory
    for root, dirs, files in os.walk(dataset_path):
        for file in files:
            if file.endswith(('.jpg', '.jpeg', '.png', '.JPG')):  # Consider image files
                total_images += 1
                
                # Extract the relative path and split into parts
                relative_path = os.path.relpath(root, dataset_path)
                parts = relative_path.split(os.sep)
                
                # Determine plant name based on folder structure
                if len(parts) == 1:
                    folder_name = parts[0]  # dataset_path/Plant_Disease
                elif len(parts) > 1:
                    folder_name = parts[-2]  # dataset_path/Plant/Disease
                else:
                    continue
                
                # Extract only the plant name
                plant_name = folder_name.split('_')[0]
                plant_counts[plant_name] += 1
    
    # Convert counts to percentages and sort in ascending order
    plant_percentages = {plant: (count / total_images) * 100 for plant, count in plant_counts.items()}
    sorted_plant_percentages = dict(sorted(plant_percentages.items(), key=lambda item: item[1]))
    
    return sorted_plant_percentages

def plot_pie_chart(plant_counts):
    labels = list(plant_counts.keys())
    sizes = list(plant_counts.values())
    
    plt.figure(figsize=(10, 7))
    plt.pie(
        sizes, labels=labels, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors,
        pctdistance=0.85, labeldistance=1.1
    )
    plt.title("Plant Distribution in Dataset")
    plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
    plt.show()

def plot_bar_chart(plant_counts):
    labels = list(plant_counts.keys())
    sizes = list(plant_counts.values())
    
    plt.figure(figsize=(12, 6))
    y_pos = range(len(labels))
    plt.barh(y_pos, sizes, color=plt.cm.Paired.colors[:len(labels)])
    plt.yticks(y_pos, labels)
    plt.xlabel("Percentage of Images")
    plt.ylabel("Plant Type")
    plt.title("Plant Distribution in Dataset (Sorted by Percentage)")
    plt.gca().invert_yaxis()  # Invert y-axis for better readability
    plt.show()
```

```python
#All_Datasets/Plant_leaf_diseases_dataset_with_augmentation/Plant_leave_diseases_dataset_with_augmentation
if __name__ == "__main__":
    dataset_path = "../All_Datasets/PlantVillage (Balanced03)"  # Change this to your local dataset path
    sorted_plant_percentages = get_plant_distribution(dataset_path)
    
    # Print plant percentages in ascending order
    print("Plant Distribution (Ascending Order by Percentage):")
    for plant, percentage in sorted_plant_percentages.items():
        print(f"{plant}: {percentage:.2f}%")
    
    plot_pie_chart(sorted_plant_percentages)
    plot_bar_chart(sorted_plant_percentages)
```
